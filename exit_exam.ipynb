```python
onehot_cols = ['company_type', 'revenue_year', 'net_income_year']
numeric_cols = ['employees', 'net_income', 'market_cap', 'direct_subsidiaries_count']
onehot_metadata = {}   # col -> list of dummy column names
X_train_enc = X_train.copy()
X_test_enc = X_test.copy()

for col in onehot_cols:
    if col not in X_train_enc.columns:
        print(f'Warning: column {col} not in dataframe; skipping one-hot for it.')
        continue
    # force string to avoid NaN category issues
    X_train_enc[col] = X_train_enc[col].astype(str)
    X_test_enc[col]  = X_test_enc[col].astype(str)

    dummies_train = pd.get_dummies(X_train_enc[col], prefix=col)
    dummy_cols = dummies_train.columns.tolist()
    onehot_metadata[col] = dummy_cols

    # attach to train
    X_train_enc = pd.concat([X_train_enc.drop(columns=[col]), dummies_train], axis=1)

    # build test dummies and align to train dummy columns
    dummies_test = pd.get_dummies(X_test_enc[col], prefix=col)
    dummies_test = dummies_test.reindex(columns=dummy_cols, fill_value=0)
    X_test_enc = pd.concat([X_test_enc.drop(columns=[col]), dummies_test], axis=1)

# 3) Median-impute then StandardScale numeric columns (fit on train only)
imputer_dict = {}
scaler_dict = {}
X_train_scaled = X_train_enc.copy()
X_test_scaled = X_test_enc.copy()

for col in numeric_cols:
    if col not in X_train_scaled.columns:
        print(f'Warning: numeric column {col} not in dataframe; skipping scaling for it.')
        continue
    # reshape for sklearn
    imputer = SimpleImputer(strategy='median')
    imputer.fit(X_train_scaled[[col]])
    imputer_dict[col] = imputer

    X_train_num = imputer.transform(X_train_scaled[[col]])
    X_test_num  = imputer.transform(X_test_scaled[[col]])

    scaler = StandardScaler()
    scaler.fit(X_train_num)
    scaler_dict[col] = scaler

    X_train_scaled[col] = scaler.transform(X_train_num).flatten()
    X_test_scaled[col]  = scaler.transform(X_test_num).flatten()

# 4) Attach targets back if present
if y is not None:
    train_df = X_train_scaled.copy()
    train_df[target] = y_train.values
    test_df = X_test_scaled.copy()
    test_df[target] = y_test.values
else:
    train_df = X_train_scaled.copy()
    test_df = X_test_scaled.copy()

# 5) Save transformed CSVs and pickles to /content
train_path = '/content/G3_train_transformed.csv'
test_path  = '/content/G3_test_transformed.csv'
train_df.to_csv(train_path, index=False)
test_df.to_csv(test_path, index=False)

# Save pickles: imputer and scaler instances, and onehot metadata
with open('/content/onehot_metadata.pkl', 'wb') as f:
    pickle.dump(onehot_metadata, f)

with open('/content/imputers.pkl', 'wb') as f:
    pickle.dump(imputer_dict, f)

with open('/content/scalers.pkl', 'wb') as f:
    pickle.dump(scaler_dict, f)

print('Saved:')
print(' -', train_path)
print(' -', test_path)
print(' - /content/onehot_metadata.pkl')
print(' - /content/imputers.pkl')
print(' - /content/scalers.pkl')
```

```python
# Train RandomForest, evaluate, save model (Colab)
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
import pickle

# load transformed train/test if variables not present
try:
    train_df
    test_df
except NameError:
    train_df = pd.read_csv('/content/G3_train_transformed.csv')
    test_df  = pd.read_csv('/content/G3_test_transformed.csv')

target = 'revenue'

X_train = train_df.drop(columns=[target])
y_train = train_df[target].astype(float)
X_test  = test_df.drop(columns=[target])
y_test  = test_df[target].astype(float)

# Train
rf = RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1)
rf.fit(X_train, y_train)

# Predict + metrics
y_pred = rf.predict(X_test)
r2   = r2_score(y_test, y_pred)
mse  = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
mae  = mean_absolute_error(y_test, y_pred)

print(f'R2:  {r2:.4f}')
print(f'MSE: {mse:,.2f}')
print(f'RMSE:{rmse:,.2f}')
print(f'MAE: {mae:,.2f}')

# Save model + feature names to Colab workspace
with open('/content/rf_model.pkl', 'wb') as f:
    pickle.dump(rf, f)
with open('/content/feature_names.pkl', 'wb') as f:
    pickle.dump(list(X_train.columns), f)

print('Saved /content/rf_model.pkl and /content/feature_names.pkl')
```

```python
# Train 3 extra regressors and print metrics
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import GradientBoostingRegressor, HistGradientBoostingRegressor
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
import pickle, time

# load transformed data if not in memory
try:
    train_df, test_df
except NameError:
    train_df = pd.read_csv('/content/G3_train_transformed.csv')
    test_df  = pd.read_csv('/content/G3_test_transformed.csv')

target = 'revenue'
X_train = train_df.drop(columns=[target])
y_train = train_df[target].astype(float)
X_test  = test_df.drop(columns=[target])
y_test  = test_df[target].astype(float)

models = {
    'LinearRegression': LinearRegression(),
    'GradientBoosting': GradientBoostingRegressor(n_estimators=200, random_state=42),
    'HistGradientBoosting': HistGradientBoostingRegressor(max_iter=200, random_state=42)
}

results = {}
for name, model in models.items():
    t0 = time.time()
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    r2 = r2_score(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(y_test, y_pred)
    dt = time.time() - t0
    results[name] = {'r2': r2, 'mse': mse, 'rmse': rmse, 'mae': mae, 'time_s': dt}
    print(f'--- {name} ---')
    print(f'R2:  {r2:.4f}')
    print(f'MSE: {mse:,.2f}')
    print(f'RMSE:{rmse:,.2f}')
    print(f'MAE: {mae:,.2f}')
    print(f'Train time: {dt:.1f}s\n')

# Optionally save models
for name, model in models.items():
    with open(f'/content/{name}.pkl', 'wb') as f:
        pickle.dump(model, f)
print('Saved models to /content/')
```
